```python
import os

project_manager_path = '/content/TyperPy_PRO/TyperPy_PRO/TyperPy/managers/project.py'

pm_content = '''from __future__ import annotations

import os
import subprocess
from pathlib import Path

from TyperPy.templates import get_templates, TemplateType


class ProjectManager:
    def __init__(self, project_name: str, template_type: str | TemplateType = TemplateType.BASIC):
        self.project_name = project_name
        self.template_type = template_type
        self.project_dir = Path(os.getcwd()) / self.project_name

    def create_structure(self) -> None:
        """Create the project structure and write template files."""
        self.project_dir.mkdir(parents=True, exist_ok=True)

        templates = get_templates(self.template_type, self.project_name)

        for template_file in templates:
            file_path = self.project_dir / template_file.name
            # Criar pastas pai se não existirem (ex: src/)
            file_path.parent.mkdir(parents=True, exist_ok=True)

            with open(file_path, "w", encoding="utf-8") as f:
                f.write(template_file.content)

        print(f"\\n✨ Project {self.project_name} created successfully in {self.project_dir}")

    def initialize_git(self) -> None:
        """Initialize a git repository in the project directory."""
        try:
            subprocess.run(["git", "init"], cwd=self.project_dir, check=True, capture_output=True)
            print("✅ Git repository initialized.")
        except subprocess.CalledProcessError:
            print("⚠️ Failed to initialize Git repository. Is git installed?")
        except FileNotFoundError:
            print("⚠️ Git command not found. Skipping git initialization.")
'''

with open(project_manager_path, 'w', encoding='utf-8') as f:
    f.write(pm_content)

print("✅ ProjectManager atualizado e corrigido para consumir o novo sistema de templates.")
```

```python
import os

doctor_manager_path = '/content/TyperPy_PRO/TyperPy_PRO/TyperPy/managers/doctor.py'

doctor_content = '''from __future__ import annotations

import shutil
import subprocess
from dataclasses import dataclass


@dataclass
class DiagnosticResult:
    tool: str
    installed: bool
    version: str | None = None
    message: str = ""


class DoctorManager:
    """Checks if the required system tools are available."""

    def check_all(self) -> list[DiagnosticResult]:
        results = []

        # 1. Verificar Git
        git_path = shutil.which("git")
        if git_path:
            try:
                version_out = subprocess.run(["git", "--version"], capture_output=True, text=True, check=True).stdout.strip()
                version = version_out.replace("git version ", "")
                results.append(DiagnosticResult("Git", True, version))
            except subprocess.CalledProcessError:
                results.append(DiagnosticResult("Git", False, message="Error reading Git version."))
        else:
            results.append(DiagnosticResult("Git", False, message="Git is not installed or not in PATH."))

        # 2. Verificar Python
        python_path = shutil.which("python") or shutil.which("python3")
        if python_path:
            try:
                version_out = subprocess.run([python_path, "--version"], capture_output=True, text=True, check=True).stdout.strip()
                version = version_out.replace("Python ", "")
                results.append(DiagnosticResult("Python", True, version))
            except subprocess.CalledProcessError:
                results.append(DiagnosticResult("Python", False, message="Error reading Python version."))
        else:
            results.append(DiagnosticResult("Python", False, message="Python is not installed or not in PATH."))

        # 3. Verificar Pip
        pip_path = shutil.which("pip") or shutil.which("pip3")
        if pip_path:
            try:
                version_out = subprocess.run([pip_path, "--version"], capture_output=True, text=True, check=True).stdout.strip()
                version = version_out.split()[1] if version_out else "Unknown"
                results.append(DiagnosticResult("Pip", True, version))
            except subprocess.CalledProcessError:
                results.append(DiagnosticResult("Pip", False, message="Error reading Pip version."))
        else:
            results.append(DiagnosticResult("Pip", False, message="Pip is not installed or not in PATH."))

        return results
'''

with open(doctor_manager_path, 'w', encoding='utf-8') as f:
    f.write(doctor_content)

print("✅ TyperPy/managers/doctor.py criado com sucesso!")
```

```python
import os

package_manager_path = '/content/TyperPy_PRO/TyperPy_PRO/TyperPy/managers/package.py'

package_content = '''from __future__ import annotations

import subprocess
import sys
from dataclasses import dataclass
from pathlib import Path


@dataclass
class PackageResult:
    success: bool
    package: str
    stdout: str = ""
    stderr: str = ""
    return_code: int = 0


class PackageManager:
    """Manages Python packages using pip."""

    def __init__(self, env_path: Path | None = None):
        self.env_path = env_path
        self._pip_cmd = self._get_pip_command()

    def _get_pip_command(self) -> list[str]:
        """Determine the correct pip command based on the environment."""
        if self.env_path:
            # If a specific environment path is provided, use its pip
            if sys.platform == "win32":
                pip_path = self.env_path / "Scripts" / "pip.exe"
            else:
                pip_path = self.env_path / "bin" / "pip"

            if pip_path.exists():
                return [str(pip_path)]

        # Fallback to current Python's pip to avoid path mismatches
        return [sys.executable, "-m", "pip"]

    def install(self, package: str) -> PackageResult:
        """Install a Python package."""
        if not package or not package.strip():
            raise ValueError("Package name cannot be empty.")

        return self._run_pip_command(["install", package], package)

    def remove(self, package: str) -> PackageResult:
        """Remove a Python package."""
        if not package or not package.strip():
            raise ValueError("Package name cannot be empty.")

        # Use -y to automatically confirm uninstallation in CLI
        return self._run_pip_command(["uninstall", "-y", package], package)

    def _run_pip_command(self, args: list[str], package: str) -> PackageResult:
        cmd = self._pip_cmd + args
        try:
            result = subprocess.run(
                cmd,
                capture_output=True,
                text=True,
                check=False
            )
            return PackageResult(
                success=(result.returncode == 0),
                package=package,
                stdout=result.stdout,
                stderr=result.stderr,
                return_code=result.returncode
            )
        except FileNotFoundError:
            return PackageResult(
                success=False,
                package=package,
                stderr=f"Could not find pip executable. Command run: {' '.join(cmd)}",
                return_code=1
            )
        except Exception as e:
            return PackageResult(
                success=False,
                package=package,
                stderr=str(e),
                return_code=1
            )
'''

with open(package_manager_path, 'w', encoding='utf-8') as f:
    f.write(package_content)

print("✅ TyperPy/managers/package.py refatorado com sucesso!")
```

```python
import os

fenyx_path = '/content/TyperPy_PRO/TyperPy_PRO/TyperPy/core/fenyx.py'

fenyx_content = '''import numpy as np
import math

class FenyxCore:
    def __init__(self, seed=12345, D=32, K=64, K_ACTIVE=3, dt=0.01, g_attract=3.0, beta=0.05, gain=10.0):
        self.D = D
        self.K = K
        self.K_ACTIVE = K_ACTIVE
        self.dt = dt
        self.g_attract = g_attract
        self.beta = beta
        self.gain = gain

        np.random.seed(seed)
        self.H = np.random.randn(K, D).astype(np.float32)
        # Normaliza as linhas
        norms = np.linalg.norm(self.H, axis=1, keepdims=True)
        self.H = np.divide(self.H, norms, out=np.zeros_like(self.H), where=norms!=0)

        self.X = np.zeros((K, D), dtype=np.float32)
        self.step_counter = 0

    def key_weights(self, key):
        logits = np.dot(self.H, key)

        # Top-K Active
        topk_indices = np.argsort(logits)[-self.K_ACTIVE:]
        vals = logits[topk_indices]

        vmin = np.min(vals)
        shifted_vals = np.maximum(vals - vmin + 1e-6, 1e-6)

        w_out = np.zeros(self.K, dtype=np.float32)
        w_out[topk_indices] = shifted_vals / np.sum(shifted_vals)
        return w_out

    def transition(self, key, value):
        w = self.key_weights(key)
        inv_1pdt = 1.0 / (1.0 + self.dt)

        for j in range(self.K):
            wj = w[j]
            if wj < 1e-12:
                self.X[j] = (self.X[j] + self.dt * np.tanh(self.g_attract * self.X[j])) * inv_1pdt
            else:
                inj = wj * value
                self.X[j] = (self.X[j] + self.dt * np.tanh(self.g_attract * self.X[j]) + self.beta * np.tanh(self.gain * inj)) * inv_1pdt

                norm = np.linalg.norm(self.X[j])
                if norm > 1e-30:
                    self.X[j] /= norm
        self.step_counter += 1

    def query(self, key):
        w = self.key_weights(key)
        out = np.dot(w, self.X)
        norm = np.linalg.norm(out)
        if norm > 1e-30:
            out /= norm
        return out
'''

with open(fenyx_path, 'w', encoding='utf-8') as f:
    f.write(fenyx_content)

print("✅ FenyxCore portado para Python e salvo em TyperPy/core/fenyx.py!")
```

```python
import time
import numpy as np
import sys

if '/content/TyperPy_PRO/TyperPy_PRO' not in sys.path:
    sys.path.insert(0, '/content/TyperPy_PRO/TyperPy_PRO')

from TyperPy.core.fenyx import FenyxCore

def benchmark_fenyx(N=1000):
    print(f"\n🚀 Iniciando teste de DeepN (Benchmark FenyxCore vs Dict Padrão) com {N} iterações...\n")
    D = 32

    # Teste 1: FenyxCore
    core = FenyxCore(D=D)
    start_time = time.time()
    for _ in range(N):
        key = np.random.randn(D).astype(np.float32)
        val = np.random.randn(D).astype(np.float32)
        core.transition(key, val)
        _ = core.query(key)
    fenyx_time = time.time() - start_time

    # Teste 2: Dicionário Padrão Python (simulando cache direto)
    std_dict = {}
    start_time = time.time()
    for _ in range(N):
        # Usando tuplas pois numpy arrays não são hashable
        key = tuple(np.random.randn(D).tolist())
        val = np.random.randn(D)
        std_dict[key] = val
        _ = std_dict.get(key)
    std_time = time.time() - start_time

    print("📊 RESULTADOS DO BENCHMARK:")
    print(f"Tempo Motor FenyxCore (Python): {fenyx_time:.4f} segundos")
    print(f"Tempo Memória Padrão (Dict):    {std_time:.4f} segundos")
    print("\n💡 Nota: O Dicionário Python padrão geralmente será mais rápido para acesso exato (O(1)),")
    print("mas o FenyxCore consome tamanho de RAM constante/fixo O(K*D) enquanto a Memória Padrão cresce linearmente O(N).")

benchmark_fenyx()
```

```python
import time
import numpy as np
from TyperPy.core.fenyx import FenyxCore

class FenyxHybridMemory:
    """
    Arquitetura Híbrida:
    L1: Dicionário Python (Acesso Exato O(1), Limite de RAM configurável)
    L2: FenyxCore Motor Vetorial (Acesso Fuzzy/Generalizado, Tamanho Fixo)
    """
    def __init__(self, D=32, K=64, cache_limit=5000):
        self.core = FenyxCore(D=D, K=K)
        self.exact_cache = {}
        self.cache_limit = cache_limit
        self.D = D

    def store(self, key: np.ndarray, value: np.ndarray):
        # Treina o motor neural profundo (sempre)
        self.core.transition(key, value)

        # Armazena no cache rápido se houver espaço na RAM simulada
        if len(self.exact_cache) < self.cache_limit:
            self.exact_cache[tuple(key.tolist())] = value

    def retrieve(self, key: np.ndarray):
        tuple_key = tuple(key.tolist())

        # 1. Tentativa L1 (Rápida e Exata)
        if tuple_key in self.exact_cache:
            return self.exact_cache[tuple_key], "L1_CACHE"

        # 2. Tentativa L2 (Fallback Vetorial)
        return self.core.query(key), "L2_FENYX"

print("✅ Classe FenyxHybridMemory (Arquitetura Híbrida) criada com sucesso!")
```

```python
def benchmark_hibrido(N_train=1000, N_test=2000):
    print(f"\n🚀 Iniciando Benchmark da Arquitetura Híbrida...")
    D = 32
    h_mem = FenyxHybridMemory(D=D, cache_limit=1000) # Limita cache exato a 1000 itens

    # Gerar dados simulados
    keys = [np.random.randn(D).astype(np.float32) for _ in range(N_test)]
    vals = [np.random.randn(D).astype(np.float32) for _ in range(N_test)]

    print(f"Treinando com {N_train} padrões...")
    for i in range(N_train):
        h_mem.store(keys[i], vals[i])

    print(f"Consultando {N_test} padrões (Metade conhecidos, Metade não vistos)...")

    start_time = time.time()
    l1_hits = 0
    l2_hits = 0

    for i in range(N_test):
        # Para i < 1000, as chaves estão no Cache L1.
        # Para i >= 1000, as chaves são novas, acionando o L2 FenyxCore.
        _, origin = h_mem.retrieve(keys[i])
        if origin == "L1_CACHE":
            l1_hits += 1
        else:
            l2_hits += 1

    total_time = time.time() - start_time

    print("\n📊 RESULTADOS DO BENCHMARK HÍBRIDO:")
    print(f"Tempo Total de Consulta: {total_time:.4f} segundos")
    print(f"Hits no Cache L1 (Exatos O(1)): {l1_hits}")
    print(f"Hits no Cache L2 (Motor Vetorial): {l2_hits}")
    print("\n💡 Conclusão: Acessos a dados conhecidos voam na velocidade da luz (Dicionário).")
    print("Acessos a dados perdidos por limite de RAM caem na rede vetorial, estabilizando uso de memória sem quebrar!")

benchmark_hibrido()
```

```python
import time
import numpy as np
import math

class FenyxCore:
    def __init__(self, seed=12345, D=32, K=64, K_ACTIVE=3, dt=0.01, g_attract=3.0, beta=0.05, gain=10.0):
        self.D = D
        self.K = K
        self.K_ACTIVE = K_ACTIVE
        self.dt = dt
        self.g_attract = g_attract
        self.beta = beta
        self.gain = gain

        np.random.seed(seed)
        self.H = np.random.randn(K, D).astype(np.float32)
        norms = np.linalg.norm(self.H, axis=1, keepdims=True)
        self.H = np.divide(self.H, norms, out=np.zeros_like(self.H), where=norms!=0)

        self.X = np.zeros((K, D), dtype=np.float32)
        self.step_counter = 0

    def key_weights(self, key):
        logits = np.dot(self.H, key)
        topk_indices = np.argsort(logits)[-self.K_ACTIVE:]
        vals = logits[topk_indices]
        vmin = np.min(vals)
        shifted_vals = np.maximum(vals - vmin + 1e-6, 1e-6)
        w_out = np.zeros(self.K, dtype=np.float32)
        w_out[topk_indices] = shifted_vals / np.sum(shifted_vals)
        return w_out

    def transition(self, key, value):
        w = self.key_weights(key)
        inv_1pdt = 1.0 / (1.0 + self.dt)
        for j in range(self.K):
            wj = w[j]
            if wj < 1e-12:
                self.X[j] = (self.X[j] + self.dt * np.tanh(self.g_attract * self.X[j])) * inv_1pdt
            else:
                inj = wj * value
                self.X[j] = (self.X[j] + self.dt * np.tanh(self.g_attract * self.X[j]) + self.beta * np.tanh(self.gain * inj)) * inv_1pdt
                norm = np.linalg.norm(self.X[j])
                if norm > 1e-30:
                    self.X[j] /= norm
        self.step_counter += 1

    def query(self, key):
        w = self.key_weights(key)
        out = np.dot(w, self.X)
        norm = np.linalg.norm(out)
        if norm > 1e-30:
            out /= norm
        return out

class FenyxHybridMemory:
    """
    Arquitetura Híbrida:
    L1: Dicionário Python (Acesso Exato O(1), Limite de RAM configurável)
    L2: FenyxCore Motor Vetorial (Acesso Fuzzy/Generalizado, Tamanho Fixo)
    """
    def __init__(self, D=32, K=64, cache_limit=5000):
        self.core = FenyxCore(D=D, K=K)
        self.exact_cache = {}
        self.cache_limit = cache_limit
        self.D = D

    def store(self, key: np.ndarray, value: np.ndarray):
        self.core.transition(key, value)
        if len(self.exact_cache) < self.cache_limit:
            self.exact_cache[tuple(key.tolist())] = value

    def retrieve(self, key: np.ndarray):
        tuple_key = tuple(key.tolist())
        if tuple_key in self.exact_cache:
            return self.exact_cache[tuple_key], "L1_CACHE"
        return self.core.query(key), "L2_FENYX"

def benchmark_hibrido(N_train=1000, N_test=2000):
    print(f"\n🚀 Iniciando Benchmark da Arquitetura Híbrida...")
    D = 32
    h_mem = FenyxHybridMemory(D=D, cache_limit=1000)

    keys = [np.random.randn(D).astype(np.float32) for _ in range(N_test)]
    vals = [np.random.randn(D).astype(np.float32) for _ in range(N_test)]

    print(f"Treinando com {N_train} padrões...")
    for i in range(N_train):
        h_mem.store(keys[i], vals[i])

    print(f"Consultando {N_test} padrões (Metade conhecidos, Metade não vistos)...")

    start_time = time.time()
    l1_hits = 0
    l2_hits = 0

    for i in range(N_test):
        _, origin = h_mem.retrieve(keys[i])
        if origin == "L1_CACHE":
            l1_hits += 1
        else:
            l2_hits += 1

    total_time = time.time() - start_time

    print("\n📊 RESULTADOS DO BENCHMARK HÍBRIDO:")
    print(f"Tempo Total de Consulta: {total_time:.4f} segundos")
    print(f"Hits no Cache L1 (Exatos O(1)): {l1_hits}")
    print(f"Hits no Cache L2 (Motor Vetorial): {l2_hits}")
    print("\n💡 Conclusão: Acessos a dados conhecidos voam na velocidade da luz (Dicionário).")
    print("Acessos a dados perdidos por limite de RAM caem na rede vetorial, estabilizando uso de memória sem quebrar!")

benchmark_hibrido(N_train=1000, N_test=2000)
```
